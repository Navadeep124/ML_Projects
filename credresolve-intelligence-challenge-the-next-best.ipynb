```python
import json

ai_call["start_time"] = pd.to_datetime(ai_call["start_time"])
ai_call["duration"] = ai_call["duration"].fillna(0)

def parse_transcript(row):
    try:
        d = json.loads(row)
        return pd.Series({
            "intent": d.get("intent"),
            "sentiment": d.get("sentiment")
        })
    except:
        return pd.Series({"intent": None, "sentiment": None})

ai_call[["intent", "sentiment"]] = ai_call["transcript_json"].apply(parse_transcript)
```

```python
whatsapp["sent_at"]=pd.to_datetime(whatsapp["sent_at"])
whatsapp["read_at"]=pd.to_datetime(whatsapp["read_at"])

whatsapp["is_read"]=whatsapp['read_at'].notna().astype(int)
whatsapp["has_response"]=whatsapp["response_message"].notna().astype(int)

wa_agg = whatsapp.groupby("lead_code").agg(
    wa_msg_sent_count=("sent_at", "count"),
    wa_msg_read_count=("is_read", "sum"),
    wa_any_response=("has_response", "max"),
    wa_num_responses=("has_response", "sum"),
    wa_last_seen=("sent_at", "max"),
)

wa_agg['wa_read_ratio']=wa_agg["wa_msg_read_count"]/wa_agg["wa_msg_sent_count"]
wa_agg["wa_read_ratio"]=wa_agg["wa_read_ratio"].fillna(0)
```

```python
ai_call["start_time"] = pd.to_datetime(ai_call["start_time"])

ai_call["is_answered"] = (ai_call["disposition"] == "ANSWERED").astype(int)
ai_call["is_busy"] = (ai_call["disposition"] == "BUSY").astype(int)
ai_call["is_noanswer"] = (ai_call["disposition"] == "NO_ANSWER").astype(int)
ai_call["is_ptp"] = (ai_call["intent"] == "PTP").astype(int)
ai_call["is_pos_sent"] = (ai_call["sentiment"] == "positive").astype(int)

bot_agg = ai_call.groupby("lead_code").agg(
    bot_call_count=("start_time", "count"),
    bot_answered_count=("is_answered", "sum"),
    bot_busy_count=("is_busy", "sum"),
    bot_noanswer_count=("is_noanswer", "sum"),
    bot_total_duration=("duration", "sum"),
    bot_ptp_count=("is_ptp", "sum"),
    bot_positive_sentiment_count=("is_pos_sent", "sum"),
    bot_last_call_time=("start_time", "max")
)

bot_agg["bot_answer_rate"] = bot_agg["bot_answered_count"] / bot_agg["bot_call_count"]
bot_agg["bot_answer_rate"] = bot_agg["bot_answer_rate"].fillna(0)
bot_agg["bot_avg_duration"] = bot_agg["bot_total_duration"] / bot_agg["bot_call_count"]
bot_agg["bot_avg_duration"] = bot_agg["bot_avg_duration"].fillna(0)
```

```python
human = human_call.copy()

human["start_time"] = pd.to_datetime(human["start_time"], errors="coerce")

human["duration"] = human["duration"].fillna(0)

human["is_answered"] = (human["disposition"] == "ANSWERED").astype(int)
human["is_noanswer"] = (human["disposition"] == "NO_ANSWER").astype(int)
human["is_busy"] = (human["disposition"] == "BUSY").astype(int)

human_agg = human.groupby("lead_code").agg(
    human_call_count=("start_time", "count"),
    human_answered_count=("is_answered", "sum"),
    human_noanswer_count=("is_noanswer", "sum"),
    human_busy_count=("is_busy", "sum"),
    human_total_duration=("duration", "sum"),
    human_unique_agents=("agent_id", "nunique"),
    human_last_call_time=("start_time", "max"),
)

human_agg["human_avg_duration"] = (
    human_agg["human_total_duration"] / human_agg["human_call_count"]
).fillna(0)

human_agg = human_agg.fillna(0)
```

```python
agent_visit["visit_date"] = pd.to_datetime(agent_visit["visit_date"])
agent_visit["door_locked"] = (agent_visit["result"] == "DOOR_LOCKED").astype(int)
agent_visit["met_customer"] = (agent_visit["result"] == "MET_CUSTOMER").astype(int)
agent_visit["shifted"] = (agent_visit["result"] == "SHIFTED").astype(int)

field_agg = agent_visit.groupby("lead_code").agg(
    field_visit_count=("visit_date", "count"),
    field_door_locked_count=("door_locked", "sum"),
    field_met_customer_count=("met_customer", "sum"),
    field_shifted_count=("shifted", "sum"),
    field_last_visit_date=("visit_date", "max")
)
```

```python
sms = ai_sms.copy()

sms["is_delivered"] = (sms["status"] == "DELIVERED").astype(int)
sms["is_failed"] = (sms["status"] == "FAILED").astype(int)

sms_agg = sms.groupby("lead_code").agg(
    sms_count=("status", "count"),
    sms_delivered_count=("is_delivered", "sum"),
    sms_failed_count=("is_failed", "sum"),
    sms_last_status=("status", lambda x: x.iloc[-1] if len(x) > 0 else None)
)

sms_agg["sms_delivery_rate"] = (
    sms_agg["sms_delivered_count"] / sms_agg["sms_count"]
).fillna(0)

sms_agg = sms_agg.fillna(0)
```

```python
customer_360 = borrowers.set_index("lead_code")

for df in [wa_agg, bot_agg, human_agg, field_agg, sms_agg]:
    customer_360 = customer_360.join(df, how="left")

customer_360 = customer_360.reset_index()
```

```python
count_cols = [c for c in customer_360.columns if "count" in c or "duration" in c]
customer_360[count_cols] = customer_360[count_cols].fillna(0)
```

```python
# Replace NaN for count columns
for col in ["sms_count", "sms_delivered_count", "sms_failed_count"]:
    customer_360[col] = customer_360[col].fillna(0)

# Replace NaN for status column with a category
customer_360["sms_last_status"] = customer_360["sms_last_status"].fillna("NO_SMS")

# Delivery rate = delivered/sms_count, if sms_count=0, rate=0
customer_360["sms_delivery_rate"] = customer_360["sms_delivery_rate"].fillna(0)
```

```python
train_merged = train.merge(customer_360, on="lead_code", how="left")
test_merged  = test.merge(customer_360,  on="lead_code", how="left")
```

```python
X = train_merged.drop(columns=["TARGET", "id"])
y = train_merged["TARGET"]
```

```python
X.head()
```

```python
X.info()
```

```python
num_cols = X.select_dtypes(include=['float64', 'int64']).columns
X[num_cols] = X[num_cols].fillna(0)

cat_cols = X.select_dtypes(include=['object']).columns
X[cat_cols] = X[cat_cols].fillna("NONE")

dt_cols = X.select_dtypes(include=['datetime64[ns]']).columns
X[dt_cols] = X[dt_cols].fillna(pd.Timestamp("2000-01-01"))

for col in dt_cols:
    X[col] = X[col].astype('int64') // 10**9
```

```python
X.info()
```

```python
cat_cols = X.select_dtypes(include=['object']).columns.tolist()
cat_cols
```

```python
X = X.drop(columns=["lead_code"])

cat_cols = X.select_dtypes(include=['object']).columns.tolist()

cat_idx = [X.columns.get_loc(c) for c in cat_cols]

from catboost import CatBoostRegressor

model = CatBoostRegressor(
    depth=8,
    learning_rate=0.05,
    iterations=1500,
    loss_function="RMSE",  
    eval_metric="RMSE",
    verbose=200
)

# 4. Fit model
model.fit(X, y, cat_features=cat_idx)
```

```python
X_test = test_merged.drop(columns=["id", "lead_code"])
X_test.info()
```

```python
num_cols_test = X_test.select_dtypes(include=['float64', 'int64']).columns
X_test[num_cols] = X_test[num_cols_test].fillna(0)

cat_cols_test = X_test.select_dtypes(include=['object']).columns
X_test[cat_cols_test] = X_test[cat_cols_test].fillna("NONE")

dt_cols_test = X_test.select_dtypes(include=['datetime64[ns]']).columns
X_test[dt_cols_test] = X_test[dt_cols_test].fillna(pd.Timestamp("2000-01-01"))

for col in dt_cols_test:
    X_test[col] = X_test[col].astype('int64') // 10**9
```

```python
test_preds = model.predict(X_test)

submission = pd.DataFrame({
    "id": test["id"],
    "TARGET": test_preds
})

submission.to_csv("submission.csv", index=False)
```
